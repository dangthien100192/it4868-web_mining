## Pipeline Overview:
1. Load product data
2. Initialize recommender with property weights
3. Vectorize product features using TF-IDF
4. Compute similarity scores between products
5. Generate recommendations for test products

## Import Required Modules

```python
from recommender import ProductRecommender
from data_loader import load_products_from_json, load_sample_products, get_product_name
from utils import print_section, print_divider, print_recommendations
```

## Load Products Data

```python
# Load products from JSON file or use sample data
try:
    products = load_products_from_json("products.json")
    print(f"✓ Loaded {len(products)} products from products.json")
except FileNotFoundError:
    print("⚠ products.json not found, using sample data...")
    products = load_sample_products()
    print(f"✓ Loaded {len(products)} sample products")

# Display first product structure
print("\nFirst product structure:")
import json
print(json.dumps(products[0], indent=2, ensure_ascii=False))
```

## Initialize Recommender with Weights

```python
# Define property weights - higher weight = more important in similarity calculation
property_weights = {
    "1": 2.0,   # Chipset (most important)
    "9": 1.5,   # Display
    "23": 1.0   # Other Info
}

# Initialize the recommender
recommender = ProductRecommender(property_weights=property_weights)

# Load products into recommender
print_section("LOADING PRODUCTS INTO RECOMMENDER")
recommender.load_products(products)
print(f"✓ Loaded {len(products)} products")
print(f"✓ Product IDs: {recommender.product_ids}")
```

## Vectorize Product Features

```python
# Vectorize product features using TF-IDF
print_section("VECTORIZING PRODUCT FEATURES")
recommender.fit(max_features=50)
print(f"\n✓ Vectorization complete")
print(f"✓ Combined vector shape: {recommender.combined_vectors.shape}")
```

## Compute Product Similarity

```python
# Compute similarity matrix using cosine similarity
print_section("COMPUTING PRODUCT SIMILARITY")
recommender.compute_similarity()
print(f"\n✓ Similarity computation complete")
print(f"✓ Similarity matrix shape: {recommender.similarity_matrix.shape}")
```

## Generate and Display Recommendations

```python
# Get first product as test case
test_product_id = products[10]['product_id']
test_product_name = products[10]['product_name']

print_section(f"RECOMMENDATIONS FOR: {test_product_name}")
print(f"Product ID: {test_product_id}\n")

# Get top 5 recommendations
recommendations = recommender.get_recommendations(test_product_id, top_k=5)

# Display recommendations
print_recommendations(recommendations, products, test_product_id)
```
